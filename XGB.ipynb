```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as pyp
import xgboost as xgb
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_csv("dataXGB\\30m\\ethereum30m.csv")
train = data[0:12298]
test = data[12298:]
```

```python
trainX, trainY = train.iloc[:, :-1], train.iloc[:,-1]
testX, testY = test.iloc[:,:-1], test.iloc[:,-1]
```

```python
model = xgb.XGBClassifier(n_estimators=10000,max_depth=8, learning_rate=0.01,objective="binary:logistic", num_class=2, verbosity=1, n_jobs=4)
```

```python
model.fit(trainX,trainY)
```

```python
predictions = model.predict(testX)
```

```python
count = 0
for i in range(len(predictions)):
    if(predictions[i]==testY[12298+i]):
        count += 1
print(count/len(testY))
```

```python
n_correct = (predictions == testY).sum()
accuracy = n_correct/len(testY)
print(accuracy)
```
