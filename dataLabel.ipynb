```python
import pandas as pd
import numpy as np

dfPre = pd.read_csv('bitcoin5m.csv')
### should we forward fill???? need justification for this (or whatever we decide)
dfPre['priceUsd'] = dfPre['priceUsd'].fillna(method='pad')
df = dfPre.copy(True)

df['priceUsd'] = df.priceUsd.shift(-1)
df['priceBtc'] = df.priceBtc.shift(-1)

df = df.iloc[1:]
dfPre = dfPre.iloc[1:]
```

```python
df2 = df.assign(neutral=( ((dfPre['priceUsd'] *0.999) < df['priceUsd']) & (df['priceUsd'] < (dfPre['priceUsd'] * 1.001))))
df2 = df2.assign(positive=(df['priceUsd'] >= dfPre['priceUsd'] * 1.001))
df2 = df2.assign(negative = (df['priceUsd'] <= dfPre['priceUsd'] * 0.999))
df2= df2.iloc[:,[0,26,27,28]] 
df2 = df2.set_index('datetime')
priceCat = []


for index,row in df2.iterrows():
    if row['positive']:
        priceCat.append('positive')
    elif row['negative']:
        priceCat.append('negative')
    else:
        priceCat.append('neutral')

df['priceDirection'] = priceCat
df

```
