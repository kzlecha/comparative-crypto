```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as pyp
import xgboost as xgb
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_csv("bitcointest.csv")
train = data[0:6000]
test = data[6000:]
```

```python
trainX, trainY = train.iloc[:, 7:-1], train.iloc[:,-1]
testX, testY = test.iloc[:,7:-1], test.iloc[:,-1]
```

```python
model = xgb.XGBClassifier(n_estimators=10000,max_depth=15, learning_rate=0.1,objective="multi:softmax", num_class=3, verbosity=2, n_jobs=4)
```

```python
model.fit(trainX,trainY)
```

```python
predictions = model.predict(testX)
```

```python
count = 0
for i in range(len(predictions)):
    if(predictions[i]==testY[6000+i]):
        count += 1
print(count/len(testY))
```
